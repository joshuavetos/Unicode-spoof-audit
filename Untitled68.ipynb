```python
import unicodedata

def build_spoof_map(target_chars="Admin"):
    """Attempts to find simple homoglyphs for target characters."""
    spoof_map = {}

    # In a full-scale research project, we'd use the official Unicode CLDR confusable list.
    # For this demo, we'll scan common similar-looking blocks: Cyrillic, Greek, and Fullwidth.
    search_blocks = [
        (0x0400, 0x04FF), # Cyrillic
        (0x0370, 0x03FF), # Greek
        (0xFF00, 0xFFEF)  # Fullwidth
    ]

    print(f"Scanning for lookalikes for: {target_chars}\n")

    for target in target_chars:
        spoof_map[target] = []
        target_name = unicodedata.name(target).split()[-1] # e.g., 'LATIN CAPITAL LETTER A' -> 'A'

        for start, end in search_blocks:
            for codepoint in range(start, end + 1):
                char = chr(codepoint)
                try:
                    name = unicodedata.name(char)
                    # Simple heuristic: if the character name ends in the same letter
                    # and it's a 'CAPITAL LETTER' or 'SMALL LETTER', it's likely a homoglyph
                    if name.endswith(f" LETTER {target_name}") or name.endswith(f" LETTER {target}"):
                        spoof_map[target].append({
                            'char': char,
                            'hex': hex(codepoint),
                            'name': name
                        })
                except ValueError:
                    continue

    # Display the map
    for char, matches in spoof_map.items():
        print(f"Target: {char}")
        for m in matches:
            print(f"  -> {m['char']} ({m['hex']}) : {m['name']}")
        if not matches:
            print("  -> No simple automated matches found in selected blocks.")
        print()

    return spoof_map

spoof_data = build_spoof_map("Admin")
```

```python
import unicodedata
import re
from confusable_homoglyphs import confusables

def levenshtein_distance(s1, s2):
    if len(s1) < len(s2): return levenshtein_distance(s2, s1)
    if len(s2) == 0: return len(s1)
    previous_row = range(len(s2) + 1)
    for i, c1 in enumerate(s1):
        current_row = [i + 1]
        for j, c2 in enumerate(s2):
            insertions = previous_row[j + 1] + 1
            deletions = current_row[j] + 1
            substitutions = previous_row[j] + (c1 != c2)
            current_row.append(min(insertions, deletions, substitutions))
        previous_row = current_row
    return previous_row[-1]

def unicode_mitigation_detector(input_str, reserved_names=None, allowed_scripts=None, whitelisted_homoglyphs=None):
    """Hardened detector using UTS #39 Confusables and script enforcement."""
    if reserved_names is None: reserved_names = ["Admin", "Root", "System", "Support"]
    if allowed_scripts is None: allowed_scripts = ["Latin"]
    if whitelisted_homoglyphs is None: whitelisted_homoglyphs = []

    results = {
        "raw": input_str,
        "scripts": set(),
        "flags": [],
        "risk_score": 0,
        "status": "APPROVED"
    }

    # 1. Whole-Script Enforcement
    for char in input_str:
        if char in whitelisted_homoglyphs: continue
        name = unicodedata.name(char, "UNKNOWN")
        if "LATIN" in name: results["scripts"].add("Latin")
        elif "CYRILLIC" in name: results["scripts"].add("Cyrillic")
        else: results["scripts"].add("Other")

    illegal_scripts = [s for s in results["scripts"] if s not in allowed_scripts]
    if illegal_scripts:
        results["flags"].append(f"Blocked Script: {illegal_scripts}")
        results["risk_score"] += 100
        results["status"] = "REJECTED"

    # 2. Advanced UTS #39 Confusable Check
    is_spoof = confusables.is_confusable(input_str, preferred_aliases=['latin'])
    if is_spoof and input_str not in reserved_names:
        results["flags"].append("Visual Spoof: UTS #39 Homoglyph detected")
        results["risk_score"] += 80
        if results["status"] == "APPROVED":
            results["status"] = "PENDING_REVIEW"

    # 3. Edit Distance Proximity
    for res in reserved_names:
        dist = levenshtein_distance(input_str.lower(), res.lower())
        if 0 < dist <= 1:
            if any(c in whitelisted_homoglyphs for c in input_str):
                results["risk_score"] += 5
            else:
                results["flags"].append(f"Proximity Alert: Near reserved name '{res}'")
                results["risk_score"] += 40
                results["status"] = "REJECTED" if dist == 0 else "PENDING_REVIEW"

    return results
```

### Bulk Mitigation Analysis

```python
def run_bulk_mitigation(target_word, spoof_map):
    options = []
    for char in target_word:
        char_options = [char] + [m['char'] for m in spoof_map.get(char, [])]
        options.append(char_options)

    all_variants = ["".join(p) for p in itertools.product(*options)]

    summary = {"Safe": 0, "Low Risk": 0, "Medium Risk": 0, "High Risk": 0}
    flag_counts = {}

    for variant in all_variants:
        if variant == target_word: continue

        res = unicode_mitigation_detector(variant)
        score = res['risk_score']

        if score == 0: summary["Safe"] += 1
        elif score < 50: summary["Low Risk"] += 1
        elif score < 100: summary["Medium Risk"] += 1
        else: summary["High Risk"] += 1

        for flag in res['flags']:
            clean_flag = flag.split(':')[0]
            flag_counts[clean_flag] = flag_counts.get(clean_flag, 0) + 1

    print(f"Total Variants Analyzed: {len(all_variants) - 1}")
    print("\n--- UPDATED Risk Distribution ---")
    for level, count in summary.items():
        print(f"{level:<12}: {count}")

    print("\n--- Mitigation Triggers ---")
    for flag, count in flag_counts.items():
        print(f"{flag:<25}: {count} detections")

run_bulk_mitigation("Admin", spoof_data)
```

### Final Mitigation: The Strict Global Policy
To close the final gaps where rendering assumptions might still allow visual spoofs (e.g., characters that look identical in specific fonts), we implement a hard-coded restriction to the Basic Latin character set (ASCII 32-126).

```python
def strict_security_policy(input_str, reserved_names=None):
    if reserved_names is None: reserved_names = ["admin", "root", "system"]

    # 1. Basic Latin Enforcement (Font-Agnostic Safety)
    # This prevents any high-unicode or homoglyph bypasses entirely
    if not all(ord(c) < 128 for c in input_str):
        return {"status": "REJECTED", "reason": "Non-ASCII characters detected (Unicode visual spoof risk)"}

    # 2. Canonical Normalization for Comparison
    normalized_input = input_str.lower().strip()

    # 3. Reserved Identity Check
    if normalized_input in reserved_names:
        return {"status": "REJECTED", "reason": "Reserved identity collision"}

    return {"status": "SECURE", "reason": "Character set restricted to safe ASCII range"}

# Verification of the Strict Policy
test_cases = ["Admin", "Adm\u0430n", "\uff21dmin", "4dmin"]

print(f"{'Input':<15} | {'Status':<10} | {'Reason'}")
print("-" * 60)
for test in test_cases:
    result = strict_security_policy(test)
    print(f"{test:<15} | {result['status']:<10} | {result['reason']}")
```

### Advanced Hardening: Flexible Unicode Security
This approach allows non-ASCII characters but applies strict script-consistency and skeleton-based lookalike detection.

```python
import unicodedata
from confusable_homoglyphs import confusables

def hardened_unicode_detector(input_str, reserved_names=None):
    """
    Final Hardened Detector: Properly treats symbols and emojis as script-neutral
    to allow international names with decorative elements.
    """
    if reserved_names is None: reserved_names = ["admin", "root", "system"]
    reserved_set = {res.lower() for res in reserved_names}

    # 1. Block Invisible/BiDi Spoofs
    bidi_chars = ["\u202a", "\u202b", "\u202c", "\u202d", "\u202e", "\u200b", "\u200c", "\u200d"]
    if any(char in input_str for char in bidi_chars):
        return {"status": "REJECTED", "reason": "Invisible or BiDi control characters detected"}

    # 2. Script Consistency Check
    scripts = set()
    neutral_patterns = ["EMOJI", "PICTOROGRAPH", "SYMBOL", "SIGN", "MARK", "ROCKET", "FIRE", "SPARKLES", "STAR", "HEART", "GLITTER", "DINGBAT"]

    for char in input_str:
        # Skip ASCII digits and underscores as neutral
        if (char.isdigit() or char == '_') and ord(char) < 128:
            continue

        try:
            name = unicodedata.name(char)
            # Skip characters that match neutral symbol/emoji patterns
            if any(p in name for p in neutral_patterns):
                continue

            # Group CJK scripts together
            if any(term in name for term in ["CJK", "HIRAGANA", "KATAKANA", "HANGUL"]):
                scripts.add("CJK_UNIFIED")
            else:
                # Record the primary script name (e.g., 'GREEK', 'LATIN')
                scripts.add(name.split(' ')[0])
        except ValueError:
            continue

    # Only reject if more than one distinct script category is found
    if len(scripts) > 1:
        return {"status": "REJECTED", "reason": f"Mixed-script detected: {list(scripts)}"}

    # 3. Reserved and Alphanumeric Checks
    input_lower = input_str.lower()
    if input_lower in reserved_set:
        return {"status": "REJECTED", "reason": "Reserved name collision"}

    # Alphanumeric spoofing check (e.g., '4dmin')
    sub_map = str.maketrans('401357', 'aoiest')
    normalized = input_lower.translate(sub_map)
    if normalized in reserved_set:
        return {"status": "REJECTED", "reason": f"Alphanumeric spoof of reserved identity detected: {normalized}"}

    # Visual Homoglyph check using UTS #39
    is_confusable = confusables.is_confusable(input_str, preferred_aliases=['latin'])
    if is_confusable:
        for match in is_confusable:
            for h in match.get('homoglyphs', []):
                if h.get('c', '').lower() in reserved_set:
                    return {"status": "REJECTED", "reason": "Visual homoglyph spoof of reserved identity"}

    return {"status": "APPROVED", "reason": "Passes script consistency and security checks"}
```
